# Do we think the same thing is happening in all of these countries?

This is a hypothetical dataset representing pesticide residues above a safety threshold vs. percentage of organic farms in different countries. How can we tell if trends are consistent across countries?

```python
data={}
for file in Path('.').glob("*.csv"):
    data[file]=pd.read_csv(file)
    df = pd.read_csv(file)
    x, y = df['x'], df['y']
    x_mean = np.mean(x)
    y_mean = np.mean(y)
    x_std= np.std(x)
    y_std= np.std(y)
    coeff = np.polyfit(x, y, deg=1)
    corr = np.corrcoef(x, y)[0, 1]


    
    print(f"x_mean={x_mean:.3f}, y_mean={y_mean:.3f}, x_std={x_std:.3f}, y_std={y_std:.3f}, linear regression fr(x)={coeff[1]:.1f} {coeff[0]:.2f} x, r = {corr:.3f} ")
```

##### Plotting

```python
for i in data:
    plt.figure(figsize=(10,7))
    plt.scatter(data[i]["x"],data[i]["y"], label=f"{i}", linewidths=0.5, s=15, color="Lightblue", edgecolors="black")
    plt.title(f"{i}")
    plt.legend()
    plt.xlabel("safety threshold")
    plt.ylabel("percentage of organic farms")
    plt.show() 
```
